```python
layoff_quarterly = layoffs_matched.groupby(['symbol', 'year', 'quarter']).agg(
    layoff_flag=('layoff_flag', 'max'),
    total_laid_off=('total_laid_off', 'sum')
).reset_index()

layoff_quarterly.head()
```

```python
fundamentals = fundamentals.rename(columns={
    'ticker_symbol': 'symbol',
    'period_ending': 'period_ending',
    'total_revenue': 'total_revenue',
    'net_income': 'net_income',
    'total_assets': 'total_assets',
    'liabilities': 'liabilities'
})

fundamentals['year'] = fundamentals['period_ending'].dt.year
fundamentals['quarter'] = fundamentals['period_ending'].dt.quarter

fundamentals.head()
```

```python
fundamentals['profit_margin'] = fundamentals['net_income'] / fundamentals['total_revenue']
fundamentals['debt_ratio'] = fundamentals['liabilities'] / fundamentals['total_assets']

fundamentals = fundamentals.replace([np.inf, -np.inf], np.nan)

fundamentals = fundamentals.dropna(
    subset=['symbol', 'year', 'quarter', 'total_revenue', 'net_income', 'total_assets']
)

fundamentals.head()
```

```python
fundamentals = fundamentals.sort_values(['symbol', 'year', 'quarter'])

fundamentals['revenue_growth'] = fundamentals.groupby('symbol')['total_revenue'].pct_change()

fundamentals[['symbol', 'year', 'quarter', 'total_revenue', 'revenue_growth', 'profit_margin', 'debt_ratio']].head(10)
```

```python
prices = prices.dropna(subset=['symbol', 'date'])

prices['year'] = prices['date'].dt.year
prices['quarter'] = prices['date'].dt.quarter

prices = prices.sort_values(['symbol', 'date'])

prices['daily_return'] = prices.groupby('symbol')['close'].pct_change()

stock_quarterly = prices.groupby(['symbol', 'year', 'quarter']).agg(
    avg_close=('close', 'mean'),
    stock_volatility=('daily_return', 'std'),
    quarterly_return=('daily_return', 'sum')
).reset_index()

stock_quarterly.head()
```

```python
final_df = fundamentals.merge(
    stock_quarterly,
    on=['symbol', 'year', 'quarter'],
    how='left'
)

final_df.head()
```

```python
# STEP 1: Create company-level target
company_layoff_target = layoffs_matched.groupby('symbol').agg(
    layoff_flag=('layoff_flag', 'max'),
    total_laid_off=('total_laid_off', 'sum')
).reset_index()

# STEP 2: Merge fundamentals + stock
final_df = fundamentals.merge(
    stock_quarterly,
    on=['symbol', 'year', 'quarter'],
    how='left'
)

# STEP 3: Merge company-level layoffs
final_df = final_df.merge(
    company_layoff_target,
    on='symbol',
    how='left'
)

# STEP 4: Fill missing values
final_df['layoff_flag'] = final_df['layoff_flag'].fillna(0).astype(int)
final_df['total_laid_off'] = final_df['total_laid_off'].fillna(0)

# Check distribution
print(final_df['layoff_flag'].value_counts())
```

```python
final_df = final_df.sort_values(['symbol', 'year', 'quarter'])

lag_features = [
    'profit_margin',
    'debt_ratio',
    'revenue_growth',
    'stock_volatility',
    'quarterly_return'
]

for col in lag_features:
    final_df[col + '_lag1'] = final_df.groupby('symbol')[col].shift(1)

final_df.head()
```

```python
model_features = [
    'profit_margin_lag1',
    'debt_ratio_lag1',
    'revenue_growth_lag1',
    'stock_volatility_lag1',
    'quarterly_return_lag1',
    'total_revenue',
    'net_income',
    'total_assets'
]

model_df = final_df.dropna(subset=model_features + ['layoff_flag'])

X = model_df[model_features]
y = model_df['layoff_flag']

print("Final modeling dataset shape:", model_df.shape)
print("Class distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(y_train_res.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_res)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

log_model = LogisticRegression(max_iter=1000)

log_model.fit(X_train_scaled, y_train_res)

log_pred = log_model.predict(X_test_scaled)
log_prob = log_model.predict_proba(X_test_scaled)[:, 1]

log_results = {
    "Model": "Logistic Regression + SMOTE",
    "Accuracy": accuracy_score(y_test, log_pred),
    "Precision": precision_score(y_test, log_pred, zero_division=0),
    "Recall": recall_score(y_test, log_pred, zero_division=0),
    "F1": f1_score(y_test, log_pred, zero_division=0),
    "ROC_AUC": roc_auc_score(y_test, log_prob)
}

log_results
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    max_depth=5
)

rf_model.fit(X_train_res, y_train_res)

rf_pred = rf_model.predict(X_test)
rf_prob = rf_model.predict_proba(X_test)[:, 1]

rf_results = {
    "Model": "Random Forest + SMOTE",
    "Accuracy": accuracy_score(y_test, rf_pred),
    "Precision": precision_score(y_test, rf_pred, zero_division=0),
    "Recall": recall_score(y_test, rf_pred, zero_division=0),
    "F1": f1_score(y_test, rf_pred, zero_division=0),
    "ROC_AUC": roc_auc_score(y_test, rf_prob)
}

rf_results
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=3,
    random_state=42,
    eval_metric='logloss'
)

xgb_model.fit(X_train_res, y_train_res)

xgb_pred = xgb_model.predict(X_test)
xgb_prob = xgb_model.predict_proba(X_test)[:, 1]

xgb_results = {
    "Model": "XGBoost + SMOTE",
    "Accuracy": accuracy_score(y_test, xgb_pred),
    "Precision": precision_score(y_test, xgb_pred, zero_division=0),
    "Recall": recall_score(y_test, xgb_pred, zero_division=0),
    "F1": f1_score(y_test, xgb_pred, zero_division=0),
    "ROC_AUC": roc_auc_score(y_test, xgb_prob)
}

xgb_results
```

```python
results_df = pd.DataFrame([log_results, rf_results, xgb_results])
results_df
```

```python
plt.figure(figsize=(8,5))

sns.barplot(data=results_df, x='Model', y='ROC_AUC')

plt.title("Model Comparison by ROC-AUC After SMOTE")
plt.xlabel("Model")
plt.ylabel("ROC-AUC")
plt.ylim(0, 1)
plt.xticks(rotation=20)

plt.show()
```

```python
from sklearn.metrics import roc_curve

log_fpr, log_tpr, _ = roc_curve(y_test, log_prob)
rf_fpr, rf_tpr, _ = roc_curve(y_test, rf_prob)
xgb_fpr, xgb_tpr, _ = roc_curve(y_test, xgb_prob)

plt.figure(figsize=(8,6))

plt.plot(log_fpr, log_tpr, label="Logistic Regression + SMOTE")
plt.plot(rf_fpr, rf_tpr, label="Random Forest + SMOTE")
plt.plot(xgb_fpr, xgb_tpr, label="XGBoost + SMOTE")
plt.plot([0,1], [0,1], linestyle='--')

plt.title("ROC Curves After SMOTE")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()

plt.show()
```
